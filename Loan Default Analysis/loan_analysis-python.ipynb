```python
numeric_columns = [
    "Income",
    "LoanAmount",
    "Age",
    "CreditScore",
    "DTI",
    "InterestRate"
]

for column in numeric_columns:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].median())
```

```python
categorical_columns = [
    "EmploymentType",
    "LoanPurpose",
    "Education",
    "MaritalStatus"
]

for column in categorical_columns:
    if column in df.columns:
        df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
numeric_columns = [
    "Age",
    "Income",
    "LoanAmount",
    "CreditScore",
    "DTI",
    "InterestRate",
    "Default"
]

correlation = df[numeric_columns].corr()

print(correlation)


plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")

plt.show()
```

```python
def create_risk_segment(row):

    if row["CreditScore"] < 580 or row["DTI"] >= 60:
        return "High Risk"

    elif row["CreditScore"] < 670 or row["DTI"] >= 40:
        return "Medium Risk"

    else:
        return "Lower Risk"


df["Risk_Segment"] = df.apply(
    create_risk_segment,
    axis=1
)

print(df["Risk_Segment"].value_counts())
```

```python
risk_default = (
    df.groupby("Risk_Segment")["Default"]
    .mean()
    .sort_values(ascending=False)
)

print(risk_default)


plt.figure(figsize=(8, 5))

sns.barplot(
    x=risk_default.index,
    y=risk_default.values
)

plt.title("Default Rate by Risk Segment")
plt.xlabel("Risk Segment")
plt.ylabel("Default Rate")

plt.show()
```
